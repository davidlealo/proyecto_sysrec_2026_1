# GreenRecSys — Tabla Comparativa por Hardware

```python
# Subir los CSVs
from google.colab import files
import pandas as pd
import io

print("Subir todos los archivos resultados_*.csv generados por el benchmark")
uploaded = files.upload()

dfs = []
for fname, content in uploaded.items():
    df = pd.read_csv(io.BytesIO(content))
    dfs.append(df)
    print(f"  Cargado: {fname} ({len(df)} filas, hardware={df['hardware'].iloc[0]})")

all_results = pd.concat(dfs, ignore_index=True)
print(f"\nTotal filas combinadas: {len(all_results)}")
print(f"Hardwares encontrados: {sorted(all_results['hardware'].unique())}")
```

```python
# Vista rápida de los datos
print("Columnas disponibles:")
print(list(all_results.columns))
all_results.head()
```

```python
# Tabla 1: Métricas de calidad por hardware y modelo (media)
quality_metrics = ["ndcg@10", "recall@10", "mrr@10", "precision@10", "hit@10", "map@10"]
available_quality = [m for m in quality_metrics if m in all_results.columns]

tabla_calidad = (
    all_results
    .groupby(["hardware", "model"])[available_quality]
    .mean()
    .round(4)
    .reset_index()
)

print("=== Tabla 1: Métricas de calidad por Hardware y Modelo (promedio) ===")
print(tabla_calidad.to_string(index=False))
```

```python
# Tabla 2: Emisiones CO₂ y tiempo por hardware y modelo
tabla_green = (
    all_results
    .groupby(["hardware", "model"])[["co2_kg", "time_s"]]
    .agg(
        co2_mean=("co2_kg", "mean"),
        co2_total=("co2_kg", "sum"),
        time_mean_s=("time_s", "mean"),
        time_total_s=("time_s", "sum"),
    )
    .round(8)
    .reset_index()
)

print("=== Tabla 2: Emisiones CO₂ y Tiempo por Hardware y Modelo ===")
print(tabla_green.to_string(index=False))
```

```python
# Tabla 3: Resumen ejecutivo por hardware (todos los modelos juntos)
resumen_hw = (
    all_results
    .groupby("hardware")
    .agg(
        experimentos=("config", "count"),
        co2_total_kg=("co2_kg", "sum"),
        co2_media_kg=("co2_kg", "mean"),
        tiempo_total_min=("time_s", lambda x: x.sum() / 60),
        tiempo_medio_s=("time_s", "mean"),
        ndcg10_media=("ndcg@10", "mean"),
        recall10_media=("recall@10", "mean"),
    )
    .round(6)
    .reset_index()
    .sort_values("co2_total_kg")
)

print("=== Tabla 3: Resumen ejecutivo por Hardware ===")
print(resumen_hw.to_string(index=False))
```

```python
# Tabla 4: Por loss function y hardware (solo modelos con loss)
loss_df = all_results[all_results["loss"] != "N/A"].copy()

if not loss_df.empty:
    tabla_loss = (
        loss_df
        .groupby(["hardware", "loss_used"])
        .agg(
            experimentos=("config", "count"),
            co2_media_kg=("co2_kg", "mean"),
            ndcg10_media=("ndcg@10", "mean"),
            recall10_media=("recall@10", "mean"),
        )
        .round(6)
        .reset_index()
    )
    print("=== Tabla 4: Loss Function × Hardware ===")
    print(tabla_loss.to_string(index=False))
else:
    print("No hay datos de loss functions en los resultados cargados.")
```

```python
# Tabla pivote final (la más útil para el paper)
# Filas = modelo, columnas = hardware, valores = NDCG@10 | CO2

pivot_ndcg = all_results.pivot_table(
    index="model",
    columns="hardware",
    values="ndcg@10",
    aggfunc="mean",
).round(4)

pivot_co2 = all_results.pivot_table(
    index="model",
    columns="hardware",
    values="co2_kg",
    aggfunc="mean",
).round(10)

pivot_time = all_results.pivot_table(
    index="model",
    columns="hardware",
    values="time_s",
    aggfunc="mean",
).round(2)

print("=== NDCG@10 por Modelo × Hardware ===")
print(pivot_ndcg.to_string())
print()
print("=== CO₂ (kg) por Modelo × Hardware ===")
print(pivot_co2.to_string())
print()
print("=== Tiempo medio (s) por Modelo × Hardware ===")
print(pivot_time.to_string())
```

```python
# Exportar todo
from google.colab import files

# CSV con todos los datos combinados
all_results.to_csv("todos_resultados_hw.csv", index=False)

# CSV con tablas de resumen en sheets separados via Excel
with pd.ExcelWriter("tabla_comparativa_hardware.xlsx") as writer:
    resumen_hw.to_excel(writer, sheet_name="Resumen_por_HW", index=False)
    tabla_calidad.to_excel(writer, sheet_name="Calidad_HW_Modelo", index=False)
    tabla_green.to_excel(writer, sheet_name="CO2_Tiempo_HW_Modelo", index=False)
    pivot_ndcg.to_excel(writer, sheet_name="Pivot_NDCG10")
    pivot_co2.to_excel(writer, sheet_name="Pivot_CO2")
    pivot_time.to_excel(writer, sheet_name="Pivot_Tiempo")
    if not loss_df.empty:
        tabla_loss.to_excel(writer, sheet_name="Loss_x_HW", index=False)

print("Archivos generados:")
print("  - todos_resultados_hw.csv")
print("  - tabla_comparativa_hardware.xlsx")

files.download("tabla_comparativa_hardware.xlsx")
files.download("todos_resultados_hw.csv")
```
